## Conclusion
Velocity means and widths are accurately inferred from ballistic expansion. Total phase is well measured by normalized spatial state-contrast shape: count alone remains at the uniform-phase error, while shape-only PCA reaches roughly 0.05–0.06 rad. Initial position/width and the final-position-preserving hidden coordinate are not recovered beyond prior accuracy by leading contrast PCA, true-phase interactions, paired images, or supervised PLS at the current sample sizes. Regime II therefore identifies a weak available local mode, but does not make generic eta regression automatically successful. A targeted phase-marginalized likelihood/posterior for that hidden mode is the next appropriate test.
